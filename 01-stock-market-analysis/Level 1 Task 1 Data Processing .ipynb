## 1. Importing libraries

## 4. Handling Missing Values

```python
df["open"] = df["open"].fillna(df["open"].median())

df["high"] = df["high"].fillna(df["high"].median())

df["low"] = df["low"].fillna(df["low"].median())
```

## Verifying Missing Values

```python
df.isnull().sum()
```

Missing Value Treatment: The dataset contained missing values in the 'open', 'high', and 'low' price columns. Rather than removing the affected records, median imputation was applied to preserve the observations while reducing the influence of extreme stock-price values.

## 5. Checking for Duplicate Rows

```python
df.duplicated().sum()
```

Duplicate Treatment: The dataset was checked for duplicate records using the duplicated() function. The result was 0, indicating that no duplicate rows were present. Therefore, no records were removed at this stage

## 6. Standardizing Data Formats

```python
df.dtypes 
```

## Standardizing the Date Format

```python
df["date"] = pd.to_datetime(df["date"])
```

```python
df.dtypes
```

Checking of Actual Dates 

```python
df["date"].head()
```

Checking for Invalid Dates 

```python
df["date"] = pd.to_datetime(df["date"], errors="coerce")
```

This will help us convert anything that isn't a valid date into NaT (missing date), allowing us to identify problematic entries instead of crashing.

```python
df["date"].isnull().sum()
```

The `date` column was initially stored as a string data type. It was converted to the pandas datetime format to ensure consistent date handling and facilitate future time-series analysis.

## 7. Final Quality Check 

```python
print("Missing values:")
print(df.isnull().sum())

print("\nDuplicate rows:")
print(df.duplicated().sum())

print("\nData types:")
print(df.dtypes)

print("\nDataset shape:")
print(df.shape)

print("\nFirst 5 rows:")
display(df.head())
```
