## Statistcal Summary of Mean, Median and Standard Deviation.

```python
mode_values = df[['open', 'high', 'low', 'close', 'volume']].mode().iloc[0]

summary_statistics = pd.DataFrame({
    'Mean': mean_values,
    'Median': median_values,
    'Mode': mode_values,
    'Standard Deviation': std_values
})

summary_statistics
```

### Summary Statistics Findings

The summary statistics show that the mean values of the opening, high, low,
and closing prices are consistently higher than their corresponding median
values. This indicates that the price distributions are positively skewed,
which is consistent with the right-skewed histograms observed during the
distribution analysis.

The stock-price variables also show substantial variability, with standard
deviations exceeding 100. Trading volume demonstrates even greater
variability, with a standard deviation of approximately 8.23 million
compared with a mean of approximately 4.25 million. The large difference
between the mean and median volume further supports the presence of a
strongly right-skewed distribution.

## 7 Distribution Analysis

## 7.1 Histogram showing distribution of closing prices 

```python
plt.figure(figsize=(8, 6))

sns.histplot(data=df, x='close', bins=30, kde=True)

plt.title('Distribution of Closing Prices')
plt.xlabel('Closing Price')
plt.ylabel('Frequency')

plt.show()
```

## Obsevrsation:
The closing-price distribution is positively skewed, with most observations concentrated at 
relatively lower price levels and a long right tail extending toward higher values.
The positive skewness is consistent with the mean closing price (86.37) being higher than the median (64.98). 
The distribution also contains a relatively small number of observations with substantially higher closing prices.

## 8 Histogram Chart Showing All Numerical Variables

```python
df[['open', 'high', 'low', 'close', 'volume']].hist(
    figsize=(10, 8),
    bins=30
)

plt.tight_layout()
plt.show()
```

## Observation: 
The distribution analysis shows that the opening, high, low, and closing prices are positively skewed, with most observations concentrated within the lower price ranges and relatively few observations extending toward substantially higher values. Trading volume exhibits an even stronger right-skewed distribution, indicating that most observations have relatively moderate trading activity while a small number record exceptionally high volumes.

## 10 — Scatterplot Analysis

## 10.1 Opening Vs Closing 

```python
plt.figure(figsize=(10, 6))

sns.scatterplot(
    data=df,
    x='open',
    y='close'
)

plt.title('Opening Price vs Closing Price')
plt.xlabel('Opening Price')
plt.ylabel('Closing Price')

plt.show()
```

## Observation: 
The scatterplot indicates a very strong positive linear relationship between opening and closing prices. Observations are closely concentrated around an upward-sloping pattern, suggesting that stocks with higher opening prices generally also have higher closing prices

## 10.2 High Vs Closing 

```python
plt.figure(figsize=(10, 6))

sns.scatterplot(
    data=df,
    x='high',
    y='close'
)

plt.title('High Price vs Closing Price')
plt.xlabel('High Price')
plt.ylabel('Closing Price')

plt.show()
```

## Observation: 
The scatterplot demonstrates a strong positive relationship between high prices and closing prices. Higher daily high prices are generally associated with higher closing prices, with the observations closely following an upward linear pattern.

## 10. 3 Low Vs Closing 

```python
plt.figure(figsize=(10, 6))

sns.scatterplot(
    data=df,
    x='low',
    y='close'
)

plt.title('Low Price vs Closing Price')
plt.xlabel('Low Price')
plt.ylabel('Closing Price')

plt.show()
```

## Observation:
A strong positive linear relationship is observed between low prices and closing prices. The close generally increases as the low price increases, resulting in a tightly clustered upward trend.

## 10.4 Volume Vs Closing 

```python
plt.figure(figsize=(10, 6))

sns.scatterplot(
    data=df,
    x='volume',
    y='close'
)

plt.title('Trading Volume vs Closing Price')
plt.xlabel('Trading Volume')
plt.ylabel('Closing Price')

plt.show()
```

## Scatterplot Findings

The scatterplot analysis reveals very strong positive relationships between
closing price and the opening, high, and low price variables. The observations
closely follow upward-sloping linear patterns, indicating that higher values
of these price variables are generally associated with higher closing prices.

In contrast, the relationship between trading volume and closing price appears
considerably weaker and less linear. The trading-volume observations are highly
concentrated at lower values with a small number of observations extending to
very high volumes. The strength of these relationships will be quantified using
correlation analysis.

## 11. Correlation Analysis

```python
correlation_matrix = df[
    ['open', 'high', 'low', 'close', 'volume']
].corr()

correlation_matrix
```

11.1 Correlation Heatmap

```python
plt.figure(figsize=(10, 7))

sns.heatmap(
    correlation_matrix,
    annot=True,
    fmt='.2f',
    cmap='coolwarm'
)

plt.title('Correlation Matrix of Numerical Features')

plt.show()
```

### Correlation Findings

The correlation analysis reveals extremely strong positive relationships
among the opening, high, low, and closing price variables, with correlation
coefficients approximately equal to 1.00. This indicates that these stock
price variables move closely together within the dataset.

Trading volume has a weak negative correlation of approximately -0.15 with
the stock-price variables. This indicates a limited negative linear
association between trading volume and stock prices.

However, correlation measures association rather than causation. Therefore,
the observed relationships should not be interpreted as evidence that one
variable directly causes changes in another.

## 12. Time-Series Analysis

```python
df = df.sort_values('date')
```

```python
plt.figure(figsize=(12, 6))

plt.plot(df['date'], df['close'])

plt.title('Close Trend Over Time', fontsize=10)
plt.xlabel("Date", fontsize=7)
plt.ylabel("Close", fontsize=7)
plt.xticks(rotation=45)
plt.grid(True, linestyle=":", alpha=0.6)


plt.tight_layout()
```

### Time-Series Findings

The time-series visualization shows considerable variation in closing prices
throughout the observation period, with fluctuations and changes in price
levels over time.

However, the dataset contains 505 different stock symbols, and the number
of observations varies between companies. Therefore, the combined time-series
plot should not be interpreted as the performance of a single stock or as a
direct representation of the overall stock market.

The visualization is useful for observing the overall distribution of price
observations across the available dates, while individual-stock analysis
would be required to assess the performance and trend of a particular company.
